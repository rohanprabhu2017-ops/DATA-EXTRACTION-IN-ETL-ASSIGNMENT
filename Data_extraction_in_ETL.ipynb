```python
#QUESTION 1 ANSWER
Structured Data: Highly organized data in fixed formats like Relational Databases (MySQL, PostgreSQL).
Semi-Structured Data: Data that doesn't reside in a relational database but has some organizational properties like XML, JSON, or CSV files.
Unstructured Data: Data with no predefined format, such as PDFs, images, or social media posts.

#QUESTION 2 ANSWER
Data extraction is the first step of the ETL process. It involves retrieving data from various source systems (like databases or APIs) so it can be moved into a staging area for transformation. Its role is to ensure that all necessary data is gathered accurately without impacting the performance of the source systems.

#QUESTION 3 ANSWER
CSV (Comma Separated Values): Plain text format, lightweight, and faster to parse. It is preferred for large datasets because it is platform-independent.
Excel (.xlsx): A binary format that supports multiple sheets, formatting, and formulas. While user-friendly, it requires more overhead to extract and can lead to data type inconsistencies.

#QUESTION 4 AMSWER
Connection: Establish a connection using drivers like JDBC or ODBC.
Identification: Determine the specific tables and columns needed.
Querying: Execute SQL queries (e.g., SELECT) to pull data.
Capture Strategy: Decide between Full Extraction (all data) or Incremental Extraction (only new/updated data since the last run).

#QUESTION 5 ANSWER
Data Volume: Extracting massive amounts of data can strain network bandwidth and source system performance.
Data Quality: Handling missing values, inconsistent formats, or corrupted records at the source.
Security/Access: Managing permissions and ensuring data privacy (GDPR/CCPA) during the transfer.

#QUESTION 6 ANSWER
An API (Application Programming Interface) allows different software systems to communicate. In ETL, APIs enable real-time extraction by allowing the system to subscribe to data changes or request updates instantly via webhooks or REST calls, rather than waiting for scheduled batch jobs.

#QUESTION 7 ANSWER
Scalability: Ability to handle petabytes of data.
Concurrency: Multiple users/processes can access data simultaneously.
Security: Robust encryption and role-based access controls.
Reliability: High availability and ACID compliance ensure data integrity.

#QUESTION 8 ANSWER
Chunking: Read the file in smaller chunks rather than loading the entire file into RAM.
Parallel Processing: Use multi-threading to process different parts of the file simultaneously.
Schema Definition: Explicitly define data types upfront to avoid the overhead of auto-detection.
Streaming: Use stream-based parsing to process records sequentially as they are read from the disk.
```
